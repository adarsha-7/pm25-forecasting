```python
import pandas as pd

path = "../../data/raw/kathmandu_full_raw_2023_2024.csv"

data = pd.read_csv(path)
```

```python
data["time"] = pd.to_datetime(data["time"])
data = data.sort_values("time")
```

```python
import matplotlib.pyplot as plt

pollutants = [
    "pm2_5",
    "pm10",
    "carbon_monoxide",
    "nitrogen_dioxide",
    "sulphur_dioxide",
    "ozone"
]

pollutant_corr = data[pollutants].corr()["pm2_5"]

plt.figure(figsize=(8,5))

plt.bar(
    pollutant_corr.index,
    pollutant_corr.values
)

plt.title("Correlation of Pollutants with PM2.5")
plt.xlabel("Pollutants")
plt.ylabel("Correlation")

plt.grid(axis="y")
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15,10))

axes = axes.flatten()

for i, pollutant in enumerate(pollutants):

    axes[i].scatter(
        data[pollutant],
        data["pm2_5"],
        alpha=0.3
    )

    axes[i].set_xlabel(pollutant)
    axes[i].set_ylabel("pm2_5")
    axes[i].set_title(f"PM2.5 vs {pollutant}")

plt.tight_layout()
plt.show()
```
