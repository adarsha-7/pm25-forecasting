```python
import pandas as pd
```

```python
data = pd.read_csv("../../../data/raw/kathmandu_full_raw_2023_2024.csv")
```

```python
data['time'] = pd.to_datetime(data['time'])
```

### 1. Check number of rows, columns, time range, hourly coverage

```python
print(f"Shape: {data.shape}")
print(f"Columns: {list(data.columns)}")
print(f"Time range: {data['time'].min()} to {data['time'].max()}")
print(f"Total hours expected: {int((data['time'].max() - data['time'].min()).total_seconds() // 3600) + 1}")
print(f"Total hours present: {len(data)}")
```

### 2. Verify timestamp, numeric, and categorical datatypes

```python
data.info()
```

### 3. Detect missing hours, irregular intervals, duplicate timestamps

```python
total_expected = int((data['time'].max() - data['time'].min()).total_seconds() // 3600) + 1
total_present = len(data)
print(f"Missing timestamps: {total_expected - total_present}")

print(f"Duplicate timestamps: {data['time'].duplicated().sum()}")

intervals = data['time'].diff().dropna()
print(f"Min interval: {intervals.min()}")
print(f"Max interval: {intervals.max()}")
print(f"All intervals equal 1 hour: {(intervals == pd.Timedelta('1h')).all()}")
```
