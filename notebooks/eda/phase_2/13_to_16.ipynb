```python
import pandas as pd
```

```python
data = pd.read_csv("../../../data/raw/kathmandu_full_raw_2023_2024.csv")
```

```python
data['time'] = pd.to_datetime(data['time'])
```

### 13. Analyze monthly and seasonal variations

```python
import matplotlib.pyplot as plt

data['month'] = data['time'].dt.month
monthly_avg = data.groupby('month')['pm2_5'].mean()

plt.figure(figsize=(10, 4))
plt.plot(monthly_avg.index, monthly_avg.values, marker='s')
plt.title('Average PM2.5 by Month')
plt.xlabel('Month')
plt.ylabel('Average PM2.5')
plt.xticks(range(1, 13))
plt.grid(True)
plt.tight_layout()
plt.show()
```

### 14. Compute moving averages and moving volatility

```python
rolling_mean_24 = data['pm2_5'].rolling(window=24).mean()
rolling_std_24 = data['pm2_5'].rolling(window=24).std()

plt.figure(figsize=(14, 4))
plt.plot(data['time'], rolling_mean_24, label='24h Rolling Mean')
plt.plot(data['time'], rolling_std_24, label='24h Rolling Std')
plt.title('PM2.5 Rolling Mean & Std (24h window)')
plt.xlabel('Time')
plt.ylabel('PM2.5')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
rolling_mean_72 = data['pm2_5'].rolling(window=72).mean()
rolling_std_72 = data['pm2_5'].rolling(window=72).std()

plt.figure(figsize=(14, 4))
plt.plot(data['time'], rolling_mean_72, label='72h Rolling Mean')
plt.plot(data['time'], rolling_std_72, label='72h Rolling Std')
plt.title('PM2.5 Rolling Mean & Std (72h window)')
plt.xlabel('Time')
plt.ylabel('PM2.5')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
