```python
import pandas as pd

path = "../../data/raw/kathmandu_full_raw_2023_2024.csv"

data = pd.read_csv(path)
```

```python
data["time"] = pd.to_datetime(data["time"])
data = data.sort_values("time")
```

```python
import matplotlib.pyplot as plt

meteorological_features = [
    "temperature_2m",
    "relative_humidity_2m",
    "wind_speed_10m",
    "wind_direction_10m",
    "surface_pressure"
]

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for i, feature in enumerate(data[meteorological_features]):
    axes[i].hist(data[feature], bins=20)
    axes[i].set_xlabel("value")
    axes[i].set_ylabel(f"{feature}")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for i, feature in enumerate(data[meteorological_features]):
    axes[i].scatter(data[feature], data["pm2_5"], alpha=0.3)
    axes[i].set_xlabel(feature)
    axes[i].set_ylabel("pm2_5")
    axes[i].set_title(f"PM2.5 vs {feature}")


plt.tight_layout()
plt.show()
```

Safe to say no visible correlation between the target and meterological features.
