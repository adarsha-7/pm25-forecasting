```python
import matplotlib.pyplot as plt

models = ['Lag 12 hr', 'Linear Reg (No Features)', 'Linear Reg', 'Lasso', 'Random Forest', 'XGBoost', 'LSTM']

rmse = [12.5512, 14.5410, 15.6999, 14.3077, 15.8645, 16.5942, 21.7706]
mae = [8.3817, 9.6592, 10.4860, 9.4599, 9.7452, 10.3088, 13.4224]
r2 = [0.8240, 0.7635, 0.7244, 0.7711, 0.7185, 0.6921, 0.4699]
```

```python
import numpy as np

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].bar(models, rmse)
axes[0].set_title('RMSE (lower is better)')
axes[0].tick_params(axis='x', rotation=60)
axes[0].set_yticks(np.arange(0, 24, 2))
axes[0].grid(True, axis='y')

axes[1].bar(models, mae)
axes[1].set_title('MAE (lower is better)')
axes[1].tick_params(axis='x', rotation=60)
axes[1].set_yticks(np.arange(0, 15, 1))
axes[1].grid(True, axis='y')

axes[2].bar(models, r2)
axes[2].set_title('R² (higher is better)')
axes[2].set_ylim(0.85, 0.95)
axes[2].tick_params(axis='x', rotation=60)
axes[2].set_yticks(np.arange(0.3, 1, 0.1))
axes[2].grid(True, axis='y')

plt.tight_layout()
plt.show()
```
