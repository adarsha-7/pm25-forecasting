```python
import matplotlib.pyplot as plt

models = ['Current PM2.5', 'Linear Reg (No Features)', 'Linear Reg', 'Lasso', 'Random Forest', 'XGBoost', 'LSTM', 'ARIMA', 'SARIMA']

rmse = [12.5561, 13.8275, 17.7788, 15.1641, 18.8529, 20.9539, 23.5018, 17.8793, 15.0272]
mae = [8.3856, 8.9266, 11.1016, 10.0611, 11.5353, 12.9652, 14.8205, 11.9491, 10.2246]
r2 = [0.8239, 0.7862, 0.6466, 0.7429, 0.6026, 0.5091, 0.3825, 0.6427, 0.7476]
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sort models from worst to best based on RMSE
order = np.argsort(rmse)[::-1]

models_sorted = [models[i] for i in order]
rmse_sorted = [rmse[i] for i in order]
mae_sorted = [mae[i] for i in order]

x = np.arange(len(models_sorted))
width = 0.38

plt.figure(figsize=(11, 5))

plt.bar(x - width/2, rmse_sorted, width, label='RMSE')
plt.bar(x + width/2, mae_sorted, width, label='MAE')

plt.title('RMSE and MAE Comparison for 24-Hour Prediction (Lower is Better)')
plt.ylabel('Error')
plt.xticks(x, models_sorted, rotation=35)
plt.yticks(np.arange(0, 26, 2))
plt.ylim(0, 25)

plt.grid(axis='y')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sort models from least to most R²
order = np.argsort(r2)

models_sorted = [models[i] for i in order]
r2_sorted = [r2[i] for i in order]

x = np.arange(len(models_sorted))

plt.figure(figsize=(11, 5))

plt.bar(x, r2_sorted, label='R²')

plt.title('R² Comparison for 24-Hour Prediction (Higher is Better)')
plt.ylabel('R²')
plt.xticks(x, models_sorted, rotation=35)
plt.yticks(np.arange(0.35, 0.86, 0.05))
plt.ylim(0.35, 0.85)

plt.grid(axis='y')
plt.legend()

plt.tight_layout()
plt.show()
```
