```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score
import time
```

```python
X_train = pd.read_csv("../../../data/processed/12_hour_prediction/X_train_linear.csv")
X_test  = pd.read_csv("../../../data/processed/12_hour_prediction/X_test_linear.csv")
y_train = pd.read_csv("../../../data/processed/12_hour_prediction/y_train.csv").squeeze()
y_test  = pd.read_csv("../../../data/processed/12_hour_prediction/y_test.csv").squeeze()
```

```python
lr = LinearRegression(n_jobs=-1)

start = time.time()
lr.fit(X_train, y_train)
training_time = time.time() - start
```

```python
start = time.time()
y_pred = lr.predict(X_test)
inference_time = (time.time() - start) / len(X_test)
```

```python
rmse = root_mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2  = r2_score(y_test, y_pred)
```

```python
print(f"RMSE          : {rmse:.4f}")
print(f"MAE           : {mae:.4f}")
print(f"R²            : {r2:.4f}")
print(f"Training time : {training_time:.2f}s")
print(f"Inference time: {inference_time*1000_000:.4f}µs")
```
