```python
import matplotlib.pyplot as plt

models = ['Lag 1 hr', 'Linear Reg', 'ARIMA', 'SARIMA']

rmse = [7.9040, 6.6911, 7.2544, 4.7298]
mae = [3.8815, 3.6420, 3.2603, 2.4223]
r2 = [0.9301, 0.9499, 0.9411, 0.9750]

# Same order as models, excluding Lag 1 hr and Linear Reg
training_times = [82.87, 1024.34]

# Same order as models, excluding Lag 1 hr and Linear Reg
inference_times = [6192.4458, 7518.7683]

# Same order as models, excluding Lag 1 hr and Linear Reg
inference_with_update = [2.425, 9.316]
```

```python
import numpy as np

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].bar(models, rmse)
axes[0].set_title('RMSE (lower is better)')
axes[0].tick_params(axis='x', rotation=15)
axes[0].set_yticks(np.arange(0, 11, 1))
axes[0].grid(True, axis='y')

axes[1].bar(models, mae)
axes[1].set_title('MAE (lower is better)')
axes[1].tick_params(axis='x', rotation=15)
axes[1].set_yticks(np.arange(0, 5, 0.5))
axes[1].grid(True, axis='y')

axes[2].bar(models, r2)
axes[2].set_title('R² (higher is better)')
axes[2].tick_params(axis='x', rotation=15)
axes[2].set_ylim((0.9,1))
axes[2].set_yticks(np.arange(0.9, 1, 0.01))
axes[2].grid(True, axis='y')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].bar(models[2:], training_times)
axes[0].set_title('Training Time (lower is better)')
axes[0].set_ylabel('Seconds')
axes[0].tick_params(axis='x', rotation=30)
axes[0].set_yticks(np.arange(0, 1100, 100))
axes[0].set_ylim(0, 1100)
axes[0].grid(True, axis='y')

axes[1].bar(models[2:], inference_times)
axes[1].set_title('Inference Time per Sample (lower is better)')
axes[1].set_ylabel('Microseconds (µs)')
axes[1].tick_params(axis='x', rotation=20)
axes[1].set_yticks(np.arange(0, 8000, 1000))
axes[1].grid(True, axis='y')

axes[2].bar(models[2:], inference_with_update)
axes[2].set_title('Inference With Update Time (lower is better)')
axes[2].set_ylabel('Seconds')
axes[2].tick_params(axis='x', rotation=20)
axes[2].set_yticks(np.arange(0, 10, 1))
axes[2].grid(True, axis='y')

plt.tight_layout()
plt.show()
```
