```python
import pandas as pd
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score
import time
```

```python
y_test  = pd.read_csv("../../../data/processed/exp_1/y_test.csv").squeeze()
```

```python
y_lag_1 = y_test.shift(1)
rmse = root_mean_squared_error(y_test[1:], y_lag_1[1:])
mae  = mean_absolute_error(y_test[1:], y_lag_1[1:])
r2   = r2_score(y_test[1:], y_lag_1[1:])

print(f"RMSE          : {rmse:.4f}")
print(f"MAE           : {mae:.4f}")
print(f"R²            : {r2:.4f}")
```

```python
y_lag_24 = y_test.shift(24)
rmse = root_mean_squared_error(y_test[24:], y_lag_1[24:])
mae  = mean_absolute_error(y_test[24:], y_lag_1[24:])
r2   = r2_score(y_test[24:], y_lag_1[24:])

print(f"RMSE          : {rmse:.4f}")
print(f"MAE           : {mae:.4f}")
print(f"R²            : {r2:.4f}")
```
