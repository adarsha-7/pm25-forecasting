```python
import matplotlib.pyplot as plt

models = ['Linear Reg', 'SVR', 'Decision Tree', 'Random Forest', 'XGBoost']
rmse  = [6.7695, 6.9474, 10.4001, 8.8816, 8.7818]
mae   = [3.6976, 3.0169, 4.8170,  4.1286, 4.4337]
r2    = [0.9487, 0.9460, 0.8790,  0.9118, 0.9137]
times = [0.14,   16.70,  1.27,    5.38,   3.42]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].bar(models, rmse)
axes[0].set_title('RMSE (lower is better)')
axes[0].tick_params(axis='x', rotation=15)

axes[1].bar(models, mae)
axes[1].set_title('MAE (lower is better)')
axes[1].tick_params(axis='x', rotation=15)

axes[2].bar(models, r2)
axes[2].set_title('R² (higher is better)')
axes[2].tick_params(axis='x', rotation=15)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
plt.bar(models, times)
plt.title('Training Time (lower is better)')
plt.ylabel('Seconds')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```
