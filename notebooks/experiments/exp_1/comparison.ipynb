```python
import matplotlib.pyplot as plt

models          = ['Lag 1 hr', 'Lag 24 hr', 'Linear Reg', 'SVR', 'Decision Tree', 'Random Forest', 'XGBoost', 'LSTM']
rmse            = [ 7.9040, 7.9284, 6.7695, 6.9620, 10.4001, 8.8816, 8.7818, 7.9390]
mae             = [3.8815, 3.8999, 3.6976, 3.0000,  4.8170,  4.1286, 4.4337, 3.8478]
r2              = [0.9301, 0.9298, 0.9487, 0.9458,  0.8790,  0.9118, 0.9137, 0.9295]
training_times  = [0.09,  15.89,    1.08,    5.13,   2.33, 34.9]
inference_times = [1.79, 1.54,  1.71, 24.62, 5.09, 17.36]
```

```python
import numpy as np

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].bar(models, rmse)
axes[0].set_title('RMSE (lower is better)')
axes[0].tick_params(axis='x', rotation=15)
axes[0].set_yticks(np.arange(0, 12, 1))
axes[0].grid(True, axis='y')

axes[1].bar(models, mae)
axes[1].set_title('MAE (lower is better)')
axes[1].tick_params(axis='x', rotation=15)
axes[1].set_yticks(np.arange(0, 5, 0.5))
axes[1].grid(True, axis='y')

axes[2].bar(models, r2)
axes[2].set_title('R² (higher is better)')
axes[2].set_ylim(0.85, 0.95)
axes[2].tick_params(axis='x', rotation=15)
axes[2].set_yticks(np.arange(0.85, 1, 0.01))
axes[2].grid(True, axis='y')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].bar(models[2:8], training_times)
axes[0].set_title('Training Time (lower is better)')
axes[0].set_ylabel('Seconds')
axes[0].tick_params(axis='x', rotation=30)
axes[0].set_yticks(np.arange(0, 36, 2))
axes[0].grid(True, axis='y')

axes[1].bar(models[2:8], inference_times)
axes[1].set_title('Inference Time per Sample (lower is better)')
axes[1].set_ylabel('Microseconds (µs)')
axes[1].tick_params(axis='x', rotation=20)
axes[1].set_yticks(np.arange(0, 34, 2))
axes[1].grid(True, axis='y')

plt.tight_layout()
plt.show()
```
