```python
import pandas as pd
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score
import time
```

```python
y_test  = pd.read_csv("../../../data/processed/24_hour_prediction/y_test.csv").squeeze()
```

```python
y_current = y_test.shift(24)

rmse = root_mean_squared_error(y_test[24:], y_current[24:])
mae  = mean_absolute_error(y_test[24:], y_current[24:])
r2   = r2_score(y_test[24:], y_current[24:])

print(f"RMSE          : {rmse:.4f}")
print(f"MAE           : {mae:.4f}")
print(f"R²            : {r2:.4f}")
```
