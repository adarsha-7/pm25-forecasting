```python
import pandas as pd
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score
import time
```

```python
y_test  = pd.read_csv("../../../data/processed/72_hour_prediction/y_test.csv").squeeze()
```

```python
y_current = y_test.shift(72)

rmse = root_mean_squared_error(y_test[72:], y_current[72:])
mae  = mean_absolute_error(y_test[72:], y_current[72:])
r2   = r2_score(y_test[72:], y_current[72:])

print(f"RMSE          : {rmse:.4f}")
print(f"MAE           : {mae:.4f}")
print(f"R²            : {r2:.4f}")
```
