```python
import matplotlib.pyplot as plt

models = ['Current PM2.5', 'Linear Reg (No Features)', 'Linear Reg', 'Lasso', 'Random Forest', 'XGBoost', 'LSTM', 'ARIMA', 'SARIMA']

rmse = [20.2519, 20.9040, 26.9295, 23.6029, 24.3310, 24.6333, 27.4036, 22.6420, 28.1443]
mae = [13.5863, 13.5296, 16.7360, 15.3649, 15.6441, 16.1151, 18.4868, 15.5698, 19.1198]
r2 = [0.5442, 0.5116, 0.1894, 0.3773, 0.3383, 0.3218, 0.1607, 0.4292, 0.1180]
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sort models from worst to best based on RMSE
order = np.argsort(rmse)[::-1]

models_sorted = [models[i] for i in order]
rmse_sorted = [rmse[i] for i in order]
mae_sorted = [mae[i] for i in order]

x = np.arange(len(models_sorted))
width = 0.38

plt.figure(figsize=(11, 5))

plt.bar(x - width/2, rmse_sorted, width, label='RMSE')
plt.bar(x + width/2, mae_sorted, width, label='MAE')

plt.title('Error Metrics Comparison for 72-Hour Prediction (Lower is Better)')
plt.ylabel('Error')
plt.xticks(x, models_sorted, rotation=35)
plt.yticks(np.arange(0, 31, 2))
plt.ylim(0, 30)

plt.grid(axis='y')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sort models from least to most R²
order = np.argsort(r2)

models_sorted = [models[i] for i in order]
r2_sorted = [r2[i] for i in order]

x = np.arange(len(models_sorted))

plt.figure(figsize=(11, 5))

plt.bar(x, r2_sorted, label='R²')

plt.title('R² Comparison for 72-Hour Prediction (Higher is Better)')
plt.ylabel('R²')
plt.xticks(x, models_sorted, rotation=35)
plt.yticks(np.arange(0.15, 0.61, 0.05))
plt.ylim(0.15, 0.60)

plt.grid(axis='y')
plt.legend()

plt.tight_layout()
plt.show()
```
