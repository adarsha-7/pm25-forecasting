```python
import matplotlib.pyplot as plt

models = ['Current PM2.5', 'Linear Reg (No Features)', 'Linear Reg', 'Lasso', 'Random Forest', 'XGBoost', 'LSTM']

rmse = [12.5561, 13.8275, 17.7788, 15.1641, 18.8529, 20.9539, 23.5018]
mae = [8.3856, 8.9266, 11.1016, 10.0611, 11.5353, 12.9652, 14.8205]
r2 = [0.8239, 0.7862, 0.6466, 0.7429, 0.6026, 0.5091, 0.3825]
```

```python
import numpy as np

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].bar(models, rmse)
axes[0].set_title('RMSE (lower is better)')
axes[0].tick_params(axis='x', rotation=45)
axes[0].set_yticks(np.arange(0, 24, 2))
axes[0].grid(True, axis='y')

axes[1].bar(models, mae)
axes[1].set_title('MAE (lower is better)')
axes[1].tick_params(axis='x', rotation=45)
axes[1].set_yticks(np.arange(0, 15, 1))
axes[1].grid(True, axis='y')

axes[2].bar(models, r2)
axes[2].set_title('R² (higher is better)')
axes[2].set_ylim(0.85, 0.95)
axes[2].tick_params(axis='x', rotation=45)
axes[2].set_yticks(np.arange(0.3, 1, 0.1))
axes[2].grid(True, axis='y')

plt.tight_layout()
plt.show()
```
