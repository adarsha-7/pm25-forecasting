```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score
import time
```

```python
X_train = pd.read_csv("../../../data/processed/1_hour_prediction/X_train_tree.csv")
X_test  = pd.read_csv("../../../data/processed/1_hour_prediction/X_test_tree.csv")
y_train = pd.read_csv("../../../data/processed/1_hour_prediction/y_train.csv").squeeze()
y_test  = pd.read_csv("../../../data/processed/1_hour_prediction/y_test.csv").squeeze()
```

```python
decision_tree = DecisionTreeRegressor(
    max_depth=15,
    min_samples_leaf=5,
    random_state=42
)

start = time.time()
decision_tree.fit(X_train, y_train)
training_time = time.time() - start
```

```python
start = time.time()
y_pred = decision_tree.predict(X_test)
inference_time = (time.time() - start) / len(X_test)
 
rmse = root_mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"RMSE          : {rmse:.4f}")
print(f"MAE           : {mae:.4f}")
print(f"R²            : {r2:.4f}")
print(f"Training time : {training_time:.2f}s")
print(f"Inference time: {inference_time*1000_000:.4f}µs")
```
