```python
import matplotlib.pyplot as plt

models = ['Lag 1 hr', 'Lag 24 hr', 'Linear Reg (No Features)', 'Linear Reg', 'Lasso', 'Random Forest', 'XGBoost', 'LSTM']

rmse = [7.9040, 7.9284, 7.3632, 6.6911, 6.7130, 8.9260, 9.1540, 8.2823]
mae = [3.8815, 3.8999, 3.4691, 3.6420, 3.6135, 4.1276, 4.4535, 4.5130]
r2 = [0.9301, 0.9298, 0.9394, 0.9499, 0.9496, 0.9109, 0.9063, 0.9233]

# Same order as models, excluding Lag 1 hr, Lag 24 hr, and Linear Reg (No Features)
training_times = [0.15, 83.70, 10.07, 4.20, 136.44]
inference_times = [2.2791, 2.2826, 29.5880, 7.3171, 171.0954]
```

```python
import numpy as np

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].bar(models, rmse)
axes[0].set_title('RMSE (lower is better)')
axes[0].tick_params(axis='x', rotation=60)
axes[0].set_yticks(np.arange(0, 12, 1))
axes[0].grid(True, axis='y')

axes[1].bar(models, mae)
axes[1].set_title('MAE (lower is better)')
axes[1].tick_params(axis='x', rotation=60)
axes[1].set_yticks(np.arange(0, 5, 0.5))
axes[1].grid(True, axis='y')

axes[2].bar(models, r2)
axes[2].set_title('R² (higher is better)')
axes[2].set_ylim(0.85, 0.95)
axes[2].tick_params(axis='x', rotation=60)
axes[2].set_yticks(np.arange(0.85, 1, 0.01))
axes[2].grid(True, axis='y')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].bar(models[3:], training_times)
axes[0].set_title('Training Time (lower is better)')
axes[0].set_ylabel('Seconds')
axes[0].tick_params(axis='x', rotation=30)
axes[0].set_yticks(np.arange(0, 140, 10))
axes[0].grid(True, axis='y')

axes[1].bar(models[3:], inference_times)
axes[1].set_title('Inference Time per Sample (lower is better)')
axes[1].set_ylabel('Microseconds (µs)')
axes[1].tick_params(axis='x', rotation=20)
axes[1].set_yticks(np.arange(0, 180, 20))
axes[1].grid(True, axis='y')

plt.tight_layout()
plt.show()
```
