```python
import pandas as pd
from sklearn.linear_model import Lasso
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score
import time
```

```python
X_train = pd.read_csv("../../../data/processed/72_hour_prediction/X_train_linear.csv")
X_test  = pd.read_csv("../../../data/processed/72_hour_prediction/X_test_linear.csv")
y_train = pd.read_csv("../../../data/processed/72_hour_prediction/y_train.csv").squeeze()
y_test  = pd.read_csv("../../../data/processed/72_hour_prediction/y_test.csv").squeeze()
```

```python
import numpy as np

alphas = [0.0001, 0.001, 0.01, 0.05, 0.1, 0.25, 0.5, 1.0, 2.0, 5.0, 10.0]

results = []
for alpha in alphas:
    lasso = Lasso(alpha=alpha)
    lasso.fit(X_train, y_train)
    y_pred = lasso.predict(X_test)

    rmse = root_mean_squared_error(y_test, y_pred)
    mae  = mean_absolute_error(y_test, y_pred)
    r2   = r2_score(y_test, y_pred)

    n_nonzero = np.sum(lasso.coef_ != 0)

    results.append({
        "alpha": alpha,
        "rmse": rmse,
        "mae": mae,
        "r2": r2,
        "n_nonzero_coefs": n_nonzero,
    })
```

```python
for result in results:
    print(f"alpha: {result["alpha"]}     rmse:{result["rmse"]}     mae:{result["mae"]}     r2:{result["r2"]}    n_nonzero:{result["n_nonzero_coefs"]}")
```

```python
lasso = Lasso(alpha=2)

start = time.time()
lasso.fit(X_train, y_train)
training_time = time.time() - start
```

```python
import numpy as np

start = time.time()
y_pred = lasso.predict(X_test)
inference_time = (time.time() - start) / len(X_test)

nonzero_coeff = np.sum(lasso.coef_ != 0)

rmse = root_mean_squared_error(y_test, y_pred)
mae  = mean_absolute_error(y_test, y_pred)
r2   = r2_score(y_test, y_pred)

print(f"n             : {nonzero_coeff} / {X_train.shape[1]}")
print(f"RMSE          : {rmse:.4f}")
print(f"MAE           : {mae:.4f}")
print(f"R²            : {r2:.4f}")
print(f"Training time : {training_time:.2f}s")
print(f"Inference time: {inference_time*1000_000:.4f}µs")
```
