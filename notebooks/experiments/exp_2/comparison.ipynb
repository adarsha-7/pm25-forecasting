```python
import matplotlib.pyplot as plt

models          = ['Lag 1 hr', 'Lag 24 hr', 'Linear Reg', 'SVR', 'ARIMA', 'SARIMA', 'Prophet']
rmse            = [7.9040, 7.9284, 6.7695, 6.9521, 40.34, 36.55, 37.52]
mae             = [3.8815, 3.8999, 3.6976, 3.0115, 27.50, 23.27, 24.75]
r2              = [0.9301, 0.9298, 0.9487, 0.9459, -0.8216, -0.4955, -0.5756]
training_times  = [1.09,    5.31,  769, 2436, 19]
inference_times = [1.7896, 32.1967, 39, 81, 382]
```

```python
import numpy as np

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].bar(models, rmse)
axes[0].set_title('RMSE (lower is better)')
axes[0].tick_params(axis='x', rotation=15)
axes[0].set_yticks(np.arange(0, 45, 5))
axes[0].grid(True, axis='y')

axes[1].bar(models, mae)
axes[1].set_title('MAE (lower is better)')
axes[1].tick_params(axis='x', rotation=15)
axes[1].set_yticks(np.arange(0, 30, 5))
axes[1].grid(True, axis='y')

axes[2].bar(models, r2)
axes[2].set_title('R² (higher is better)')
axes[2].set_ylim(-1, 1)
axes[2].tick_params(axis='x', rotation=15)
axes[2].set_yticks(np.arange(-1, 1, 0.1))
axes[2].grid(True, axis='y')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].bar(models[2:7], training_times)
axes[0].set_title('Training Time (lower is better)')
axes[0].set_ylabel('Seconds')
axes[0].tick_params(axis='x', rotation=30)
axes[0].set_yticks(np.arange(0, 100, 10))
axes[0].set_ylim(0, 100)
axes[0].grid(True, axis='y')

axes[1].bar(models[2:7], inference_times)
axes[1].set_title('Inference Time per Sample (lower is better)')
axes[1].set_ylabel('Microseconds (µs)')
axes[1].tick_params(axis='x', rotation=20)
axes[1].set_yticks(np.arange(0, 400, 20))
axes[1].grid(True, axis='y')

plt.tight_layout()
plt.show()
```
