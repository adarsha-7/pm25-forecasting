```python
import matplotlib.pyplot as plt

models = ['Current PM2.5', 'Linear Reg (No Features)', 'Linear Reg', 'Lasso', 'Random Forest', 'XGBoost', 'LSTM']

recall = [0.8785, 0.9230, 0.9102, 0.9214, 0.9031, 0.9095, 0.9182]

f1 = [0.8837, 0.8850, 0.8792, 0.8845, 0.8756, 0.8787, 0.8828]
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sort models from lowest to highest F1
order = np.argsort(recall)

models_sorted = [models[i] for i in order]
recall_sorted = [recall[i] for i in order]
f1_sorted = [f1[i] for i in order]

x = np.arange(len(models_sorted))
width = 0.38

plt.figure(figsize=(11, 5))

plt.bar(x - width/2, recall_sorted, width, label='Recall')
plt.bar(x + width/2, f1_sorted, width, label='F1-score')

plt.title('Recall and F1-score Comparison for 24-Hour Classification\n(Precision ≈ 0.85)')
plt.ylabel('Score')
plt.xticks(x, models_sorted, rotation=35)
plt.yticks(np.arange(0.84, 0.94, 0.005))
plt.ylim(0.84, 0.93)

plt.grid(axis='y')
plt.legend()

plt.tight_layout()
plt.show()
```
