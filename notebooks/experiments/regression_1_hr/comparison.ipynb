```python
import matplotlib.pyplot as plt

models = ['Lag 1 hr', 'Linear Reg (No Features)', 'Linear Reg', 'Lasso', 'Random Forest', 'XGBoost', 'LSTM', 'ARIMA', 'SARIMA']

rmse = [7.9040, 7.3632, 6.6911, 6.7130, 8.9260, 9.1540, 8.2823, 7.2544, 4.7298]
mae = [3.8815, 3.4691, 3.6420, 3.6135, 4.1276, 4.4535, 4.5130, 3.2603, 2.4223]
r2 = [0.9301, 0.9394, 0.9499, 0.9496, 0.9109, 0.9063, 0.9233, 0.9411, 0.9750]
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sort models from worst to best based on RMSE
order = np.argsort(rmse)[::-1]

models_sorted = [models[i] for i in order]
rmse_sorted = [rmse[i] for i in order]
mae_sorted = [mae[i] for i in order]

x = np.arange(len(models_sorted))
width = 0.38

plt.figure(figsize=(10, 6))

plt.bar(x - width/2, rmse_sorted, width, label='RMSE')
plt.bar(x + width/2, mae_sorted, width, label='MAE')

plt.title('RMSE and MAE Comparison for 1-Hour Prediction (Lower is Better)', fontsize=15)
plt.ylabel('Error', fontsize=14)
plt.xticks(x, models_sorted, rotation=45, fontsize=14)
plt.yticks(np.arange(0, 11, 1), fontsize=14)
plt.ylim(0, 10)

plt.grid(axis='y')
plt.legend(fontsize=15)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sort models from least to most R²
order = np.argsort(r2)

models_sorted = [models[i] for i in order]
r2_sorted = [r2[i] for i in order]

x = np.arange(len(models_sorted))

plt.figure(figsize=(10, 6))

plt.bar(x, r2_sorted, label='R²')

plt.title('R² Comparison for 1-Hour Prediction (Higher is Better)', fontsize=15)
plt.ylabel('R²', fontsize=14)
plt.xticks(x, models_sorted, rotation=45, fontsize=14)
plt.yticks(np.arange(0.90, 0.981, 0.01), fontsize=14)
plt.ylim(0.90, 0.98)

plt.grid(axis='y')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

models = ['Linear Reg\n(No Features)', 'Linear Reg', 'Lasso', 'Random Forest', 'XGBoost', 'ARIMA', 'SARIMA']

inference_times = [2.2791, 2.2826, 29.5880, 7.3171, 171.0954, 6192.4458, 7518.7683]

# Sort from highest to lowest inference time
order = np.argsort(inference_times)[::-1]

models_sorted = [models[i] for i in order]
times_sorted = [inference_times[i] for i in order]

x = np.arange(len(models_sorted))

plt.figure(figsize=(10, 5))

plt.bar(x, times_sorted)

plt.title('Inference Time per Sample (Lower is Better)')
plt.ylabel('Microseconds (µs)')
plt.xticks(x, models_sorted, rotation=30)

plt.yticks(np.arange(0, 8001, 1000))
plt.ylim(0, 8000)

plt.grid(axis='y')

plt.tight_layout()
plt.show()
```
